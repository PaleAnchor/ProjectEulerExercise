```python
NEXT_ROW = 20
NEXT_WORD = 1


def index(x, y):
    target = NEXT_ROW * x + NEXT_WORD * y
    return target


with open(
    "C:/Users/lanon/OneDrive/文档/程序库/Project Euler/数据/011.txt",
    "r",
    encoding="utf-8",
) as file:
    data = file.read().split()
    sum = 0
    matrix = [int(x) for x in data]
    for i in range(0, 17):
        for j in range(0, 20):
            sum_c = (
                matrix[index(i, j)]
                * matrix[index(i + 1, j)]
                * matrix[index(i + 2, j)]
                * matrix[index(i + 3, j)]
            )
            if sum_c > sum:
                sum = sum_c
    for i in range(0, 20):
        for j in range(0, 17):
            sum_h = (
                matrix[index(i, j)]
                * matrix[index(i, j + 1)]
                * matrix[index(i, j + 2)]
                * matrix[index(i, j + 3)]
            )
            if sum_h > sum:
                sum = sum_h
    for i in range(0, 17):
        for j in range(0, 17):
            sum_x = (
                matrix[index(i, j)]
                * matrix[index(i + 1, j + 1)]
                * matrix[index(i + 2, j + 2)]
                * matrix[index(i + 3, j + 3)]
            )
            if sum_x > sum:
                sum = sum_x
    for i in range(0, 17):
        for j in range(3, 20):
            sum_ix = (
                matrix[index(i, j)]
                * matrix[index(i + 1, j - 1)]
                * matrix[index(i + 2, j - 2)]
                * matrix[index(i + 3, j - 3)]
            )
            if sum_ix > sum:
                sum = sum_ix
print(sum) 
```
