第一个约数数量超过五百的三角形数是多少？

```python
import numpy as np
import math
tri=1
loop_num=1
while(1):
    factor_num=0
    tri=tri+loop_num+1
    for i in range(1, math.ceil(np.sqrt(tri))):
        if tri % i == 0:
            factor_num = factor_num + 2
    if(factor_num>500):
        print(tri)
        break
    loop_num=loop_num+1
```

要点:math中的取整函数ceil,floor round ,以及pyhton中自带的int取整
