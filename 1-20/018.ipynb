最大路径和（一）

```python
TARGET_LINE=-2

with open("Project Euler/数据/018.txt","r") as file:
    data=[[int(i) for i in sublist.split()] for sublist in file.readlines()]
    while(len(data)>1):
        for i in range(len(data[TARGET_LINE])):
            if data[-1][i]>data[-1][i+1]:
                data[TARGET_LINE][i]=data[TARGET_LINE][i]+data[-1][i]
            else:
                data[TARGET_LINE][i]=data[TARGET_LINE][i]+data[-1][i+1]
        data=data[:-1]    

    print(data)
```
