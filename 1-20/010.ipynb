010:求所有小于两百万的质数的和

```python
#埃拉托斯特尼筛法确定某个范围内所有素数
import numpy as np
limit=2000000
num=np.ones(limit-1,dtype=bool)
min_prime_id=np.argmax(num)
min_prime=min_prime_id+2
while(min_prime<np.sqrt(len(num)+1)):
    for i in range(min_prime_id+min_prime,len(num),min_prime):
        num[i]=False
    min_prime_id=np.argmax(num[min_prime_id+1:])+len(num[:min_prime_id+1])
    min_prime=min_prime_id+2

sum=0
for i in range(0,limit-1):
    if num[i]==True:
        sum=sum+(i+2)
print(sum)
```

#### 要点:

argmax()切片:[:-1]
