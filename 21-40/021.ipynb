```python
import math
import numpy as np
def FactorSum(num:int):
    result=1
    for i in range(2, int(np.sqrt(num))+1):
        if num % i == 0:
            if i<np.sqrt(num):
                result=result+i+int(num/i)
            else:
                result=result+i
    return result

sum=0
dic={}
for i in range(2,10000):
    if i not in dic:
        dic[FactorSum(i)]=i
    else:
        if dic[i]==FactorSum(i):
            sum=sum+i+FactorSum(i)

# print(dic)
print(sum)
```
