```python
import re
import sys
sys.set_int_max_str_digits(0)  # 0 disables the limit completely

name_loc = r"C:\Users\V\Documents\欧拉计划\ProjectEulerExercise\数据\022.txt"
with open(name_loc) as f:
    content = f.read()
pattern = r'"(.*?)"'
name_list = re.findall(pattern, content)
name_list.sort()

name_value = 0
for i in range(len(name_list)):
    single_name_value = 0
    for j in range(len(name_list[i])):
        single_name_value = single_name_value + ord(name_list[i][j]) - ord("A") + 1
    single_name_value = single_name_value * (i + 1)
    name_value += single_name_value

print(name_value)
```
